# folium 으로 지도 시각화하기

```python
import folium
```

```python
m = folium.Map(location=[37.513096, 126.938555], zoom_start=12)
#m.save('map.html')
folium.Marker([37.513096, 126.938555], popup='Hi IGrowth').add_to(m)
m
```

```python
37.606340, 126.935969
```

```python
k = folium.Map(location=[37.606340, 126.935969], zoom_start=12)
#m.save('map.html')
folium.Marker([37.606340, 126.935969], popup='Home').add_to(k)
k
```

```python
from geopy.distance import geodesic

# 집과 회사 좌표 (예시)
home = (37.606340, 126.935969)      # 서울시청
office = (37.513096, 126.938555)   # 강남역

# 지도 중심은 중간 지점
m = folium.Map(location=[(home[0]+office[0])/2,
                         (home[1]+office[1])/2],
               zoom_start=12)

folium.Marker(home, tooltip="집", icon=folium.Icon(color="blue")).add_to(m)
folium.Marker(office, tooltip="회사", icon=folium.Icon(color="red")).add_to(m)

folium.PolyLine([home, office],
                color="green",
                weight=3,
                dash_array="5").add_to(m)

distance = geodesic(home, office).km
print(distance)

midpoint = [(home[0]+office[0])/2,
            (home[1]+office[1])/2]

folium.Marker(
    midpoint,
    icon=folium.DivIcon(
        html=f"<div style='font-size: 14pt'>{distance:.2f} km</div>"
    )
).add_to(m)

m.save("commute_map.html")
m
```
